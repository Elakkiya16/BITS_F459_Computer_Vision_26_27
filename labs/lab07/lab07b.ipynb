## E1 &middot; Classify Q (1 mark)

Give the label of your Q for k = 1, k = 3 and k = 5.

**What you should see.** Three labels, each `"sky"`, `"road"` or `"tree"`, that you can confirm from a

## E2 &middot; A point where one neighbour and three neighbours disagree (1 mark)

Give a point P, different from Q and from every training point, whose 1-NN label differs from
its 3-NN label.

`ANS_E2 = [x, y]`

**Worked example.** P = (2.5, 0). Nearest: #10 tree 1.50, #5 road 2.06, #6 road 2.50. 1-NN says
tree, 3-NN says road. Answer `[2.5, 0.0]`.

**What you should see.** At your point, k-NN with k = 1 and with k = 3 gives two different labels, and
the point is neither Q nor one of your 12 training points.

## E3 &middot; A point where the two rulers disagree (1 mark)

Give a point P, different from every training point, whose nearest training point under L1
has a different label from its nearest training point under L2.

`ANS_E3 = [x, y]`

**Worked example.** P = (10, 0). Under L1, #10 tree at (4, 0) is 6 away and #1 sky at (7, 5) is
3 + 5 = 8 away, so L1 says tree. Under L2, #10 is 6.00 away and #1 is &radic;(3&sup2; + 5&sup2;) = 5.83
away, so L2 says sky. Answer `[10.0, 0.0]`.

**What you should see.** At your point, the nearest training point under L1 and the nearest under L2
have different labels, and the point is not one of your 12 training points.

## E4 &middot; Add one point that changes the vote (1 mark)

Add one new labelled training point, at L2 distance **at least 2** from Q, so that the 3-NN
label of Q changes. The new point becomes point 13.

`ANS_E4 = {"point": [x, y], "label": "sky" or "road" or "tree"}`

**Worked example.** Q = (5, 0) is road with k = 3. Add a sky point at (1.5, 0), 3.50 from Q.
The three nearest become #10 tree 1.00, #6 road 2.24 and #13 sky 3.50, which pushes out #5 road
at 3.61. One vote each, so the tie rule returns the label of the nearest tied point: tree.
Answer `{"point": [1.5, 0.0], "label": "sky"}`.

**What you should see.** Your point is at least 2.00 from Q, and the 3-NN label of Q with your 13
points differs from its label with the original 12.

## E5 &middot; How much stretching changes the answer (1 mark)

Multiply the **first** coordinate of every point, Q included, by a whole number s. Give the
smallest s &ge; 2 for which the 3-NN label of Q differs from its label at s = 1.

`ANS_E5 = s`

**Worked example.** At s = 1, Q is road. At s = 2, Q becomes (10, 0) and the three nearest are
#10 tree (8, 0) 2.00, #6 road (8, 2) 2.83 and #2 sky (12, 5) 5.39. One vote each, so the tie
rule returns tree. The label has changed already at s = 2. Answer `2`.

**What you should see.** At your s the 3-NN label of Q differs from its label at s = 1, and at every
whole number from 2 up to s &minus; 1 it does not.

## E6 &middot; Remove the fewest points (1 mark)

Remove as few training points as possible so that the 3-NN label of Q changes. Give their
point numbers.

`ANS_E6 = [point numbers]`

**Worked example.** Q is road because #6 and #5 outvote #10. Removing one road point, say #6,
brings in #7, another road, so Q stays road. Removing two, #5 and #6, leaves #10 tree 1.00,
#7 road 4.24 and #2 sky 5.10: one vote each, and the tie rule returns tree. Two is the fewest.
Answer `[5, 6]`; `[5, 7]` and `[6, 7]` are also correct.

**What you should see.** Without your points the 3-NN label of Q changes, and no smaller set of
removals changes it.

## E7 &middot; Stretch &middot; Find the hidden rule (1 mark)

`mystery(P)` returns the label that a hidden k-NN gives to a point P, using **your** 12 training
points. Its k is one of 1, 3, 5, 7 and its distance is L1 or L2. Find both by testing points of
your choice.

`ANS_E7 = {"k": 1 or 3 or 5 or 7, "metric": "L1" or "L2"}`

**Worked example.** `mystery_example` hides a rule on the worked dataset.
`mystery_example((5, 0))` returns road, so k is not 1, because 1-NN gives tree there. At (3, 5)
the eight possible rules give: k = 1 tree (both rulers); k = 3 tree with L1, road with L2;
k = 5 sky with L1, road with L2; k = 7 road (both). `mystery_example((3, 5))` returns sky, so the
rule is k = 5 with L1. Answer `{"k": 5, "metric": "L1"}`.

**What you should see.** For every point you test, your rule gives the same label as `mystery`, and
no other of the eight rules agrees with `mystery` on all of your test points.
